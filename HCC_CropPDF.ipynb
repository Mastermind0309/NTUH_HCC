```python
!pip install PyPDF2

from google.colab import drive
drive.mount('/content/drive')

!apt-get -y install ghostscript

import os
from PyPDF2 import PdfReader, PdfWriter

def crop_pdf(input_path, output_path, left=10, right=10, top=10, bottom=10):
    """
    Crop white borders from a PDF (in points; 72 points = 1 inch).
    Adjust the margins (left, right, top, bottom) to fine-tune.
    """
    reader = PdfReader(input_path)
    writer = PdfWriter()

    for page in reader.pages:
        mediabox = page.mediabox
        x0, y0 = mediabox.lower_left
        x1, y1 = mediabox.upper_right

        # Shrink page by specified margins
        new_x0 = x0 + left
        new_y0 = y0 + bottom
        new_x1 = x1 - right
        new_y1 = y1 - top

        # Apply crop box
        page.mediabox.lower_left = (new_x0, new_y0)
        page.mediabox.upper_right = (new_x1, new_y1)
        page.cropbox.lower_left = (new_x0, new_y0)
        page.cropbox.upper_right = (new_x1, new_y1)

        writer.add_page(page)

    with open(output_path, "wb") as f_out:
        writer.write(f_out)
    print(f"Saved cropped PDF → {output_path}")

# ---- File paths ----
figure_pdfs = [
    "/content/drive/MyDrive/HCC/Figure1.pdf",
    "/content/drive/MyDrive/HCC/Figure2.pdf"
]

# ---- Crop parameters (tweak if needed) ----
# The values below trim ~10 pt (~3.5 mm) on each side.
for fig in figure_pdfs:
    out_path = fig.replace(".pdf", "_cropped.pdf")
    crop_pdf(fig, out_path, left=10, right=10, top=10, bottom=10)
```
