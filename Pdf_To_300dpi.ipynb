```python
from google.colab import drive
drive.mount('/content/drive')
!apt-get install poppler-utils
!pip install pdf2image
from pdf2image import convert_from_path
from PIL import Image

pdf_files = [
    "/content/drive/MyDrive/HCC/Figure3.pdf",
    # "/content/drive/MyDrive/HCC/Paper/HCC_paper_materials/oral_cancer_figure2_0919.pdf"
]

for pdf in pdf_files:
    pages = convert_from_path(pdf, dpi=600)
    for i, page in enumerate(pages):
        out_path = pdf.replace(".pdf", f"_{i+1}_600dpi.tif")
        page.save(out_path, "TIFF", dpi=(600,600))
        print("Saved:", out_path)

# Verify
img = Image.open(out_path)
print("Resolution:", img.info['dpi'])

```
