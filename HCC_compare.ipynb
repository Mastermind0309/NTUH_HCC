```python
# mount


from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import re, unicodedata

sdm_path = "/content/drive/MyDrive/HCC/SDM_2018.xlsx"
duo_path = "/content/drive/MyDrive/HCC/Duo_2018.xlsx"



sdm = pd.read_excel(sdm_path)
duo = pd.read_excel(duo_path)

ids_col_sdm = "Patient_ID1"
ids_col_duo = "Patient_ID2"

ids_sdm = set(sdm[ids_col_sdm].dropna().astype(str))
ids_duo = set(duo[ids_col_duo].dropna().astype(str))

only_in_sdm = ids_sdm - ids_duo
only_in_duo = ids_duo - ids_sdm
in_both = ids_sdm & ids_duo

print("Unique patients in SDM 2018:", len(ids_sdm))
print("Unique patients in Duo 2018:", len(ids_duo))
print("Patients in both:", len(in_both))
print("Only in SDM:", len(only_in_sdm))
print("Only in Duo:", len(only_in_duo))


all_ids = sorted(ids_sdm | ids_duo)
df_compare = pd.DataFrame({
    "PatientID": all_ids,
    "In_SDM": ["Yes" if pid in ids_sdm else "No" for pid in all_ids],
    "In_Duo": ["Yes" if pid in ids_duo else "No" for pid in all_ids],
})


output_path = "/content/drive/MyDrive/HCC/patient_id_comparison_2018.xlsx"

df_compare.to_excel(output_path, index=False)

print("Comparison saved to:", output_path)
print("Raw distinct IDs from file:", sdm[ids_col_sdm].nunique())
print("After cleaning:", len(ids_sdm))

```
