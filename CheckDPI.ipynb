```python
from PIL import Image

# ===== Step 1: Mount Google Drive (if your image is in Drive) =====
from google.colab import drive
drive.mount('/content/drive')

# ===== Step 2: Set your image file path =====
# Example: replace with your actual file path
file_path = '/content/drive/MyDrive/HCC/TIF_out/Figure2.tif'

# ===== Step 3: Open image and check DPI =====
with Image.open(file_path) as img:
    dpi = img.info.get('dpi')
    width, height = img.size
    print(f" File: {file_path}")
    print(f" Dimensions: {width} x {height} px")
    if dpi:
        print(f" DPI: {dpi[0]} x {dpi[1]}")
    else:
        print(" DPI information not found in metadata.")
```
