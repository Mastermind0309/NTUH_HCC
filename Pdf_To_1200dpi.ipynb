```python
from google.colab import drive
drive.mount('/content/drive')

!apt-get -y install poppler-utils
!pip install pdf2image Pillow

from pdf2image import convert_from_path
from PIL import Image, TiffImagePlugin

#  Allow large images (disables Pillow's decompression bomb check)
Image.MAX_IMAGE_PIXELS = None

pdf_files = [
    "/content/drive/MyDrive/HCC/Figure1.pdf",
    "/content/drive/MyDrive/HCC/Figure2.pdf",
]

for pdf in pdf_files:
    pages = convert_from_path(pdf, dpi=1200)  # 1200 DPI for line art
    for i, page in enumerate(pages, 1):
        out_path = pdf.replace(".pdf", f"_{i}_1200dpi.tif")
        # Use lossless LZW compression to keep file smaller
        page.save(out_path, format="TIFF", dpi=(1200, 1200), compression="tiff_lzw")
        print("Saved:", out_path)

# Verify the last saved file
img = Image.open(out_path)
print("Resolution:", img.info.get('dpi'))
print("Mode:", img.mode, "| Size:", img.size)
```
