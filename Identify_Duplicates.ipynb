```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import re, unicodedata

# Load your SDM file
sdm_path = "/content/drive/MyDrive/HCC/SDM_2018.xlsx"
df = pd.read_excel(sdm_path)

id_col = "Patient_ID1"   # or "Patient_ID2" for Duo

# Clean IDs a bit
df[id_col] = df[id_col].astype(str).str.strip().replace({"nan": None, "NaN": None, "None": None})

# Count how many times each ID appears
dup_counts = df[id_col].value_counts()

# Filter only IDs that appear more than once
duplicates = dup_counts[dup_counts > 1]

# Print
print("IDs that appear more than once:")
print(duplicates)

# (Optional) save to Excel
duplicates.to_excel("/content/drive/MyDrive/HCC/duplicate_IDs_in_SDM_2018.xlsx")
print("Saved duplicate list to /MyDrive/HCC/duplicate_IDs_in_SDM_2018.xlsx")
```
