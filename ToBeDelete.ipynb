```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd


# File paths
sdm_path = "/content/drive/MyDrive/HCC/SDM_2018.xlsx"
duo_path = "/content/drive/MyDrive/HCC/Duo_2018.xlsx"

# Patient ID columns
id_col_sdm = "Patient_ID1"
id_col_duo = "Patient_ID2"

# Load data
sdm = pd.read_excel(sdm_path)
duo = pd.read_excel(duo_path)

# Clean IDs
sdm[id_col_sdm] = sdm[id_col_sdm].astype(str).str.strip().replace({"nan": None, "NaN": None, "None": None})
duo[id_col_duo] = duo[id_col_duo].astype(str).str.strip().replace({"nan": None, "NaN": None, "None": None})

# Get unique IDs
ids_sdm = set(sdm[id_col_sdm].dropna())
ids_duo = set(duo[id_col_duo].dropna())

# Find IDs only in SDM
only_in_sdm = sorted(ids_sdm - ids_duo)

print("Number of IDs only in SDM:", len(only_in_sdm))

# Save to Excel
out_path = "/content/drive/MyDrive/HCC/IDs_only_in_SDM_2018.xlsx"
pd.DataFrame({"Patient_ID_only_in_SDM": only_in_sdm}).to_excel(out_path, index=False)

print("Full list saved to:", out_path)
```
