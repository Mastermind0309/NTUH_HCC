```python
# This code converts files into SPSS readable file
from google.colab import drive
drive.mount ('/content/drive')
excel_path = "/content/drive/MyDrive/HCC/HCC_SDM_20250728_revised_0829.xlsx"
output_path = "/content/drive/MyDrive/HCC/HCC_SDM_20250728_revised_cp950_0829.csv"
import pandas as pd
import re
import unicodedata


df = pd.read_excel(excel_path)

# Columns to clean
target_cols = ["壹-一-3_死亡地點", "壹-一-4_DNR"]

# Normalize and replace NBSP/zero-width spaces with regular spaces, then strip
def clean_cell(x):
    if pd.isna(x):
        return x
    s = unicodedata.normalize("NFKC", str(x))
    s = s.replace("\u00A0", " ").replace("\u200B", "")
    s = re.sub(r"[ ]{2,}", " ", s).strip()

    # Remove private use characters (like \ue09b, \uec28)
    s = re.sub(r"[\ue000-\uf8ff]", "", s)

    # Try encoding to cp950; if it fails, drop or replace bad chars
    try:
        s.encode("cp950")
    except UnicodeEncodeError:
        s = s.encode("cp950", errors="replace").decode("cp950")
    return s

# Apply to all object columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].map(clean_cell)

df.to_csv(output_path, index=False, encoding="cp950")


print("Saved:", output_path)

```
